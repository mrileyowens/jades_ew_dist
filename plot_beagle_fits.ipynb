```python
import sys

import os
import glob

sys.path.append(os.path.abspath('..'))

from mrileyowens.beagle import plot
```

```python
def plot_beagle_fits():

    home = os.getcwd()
    data = f'{home}/data'
    figs = f'{home}/figs'
    results = f'{home}/results'

    # Set the prefixes for the different sets of BEAGLE fits
    names = ['JADES_z6to9LBGcatalog_Endsley2024_f775w_dropouts_beagle_csfh_fits', 'JADES_z6to9LBGcatalog_Endsley2024_f775w_dropouts_beagle_csfh_fits_no_lya']

    for i, name in enumerate(names):

        files = glob.glob(f'{results}/beagle_fits/{name}/*_GOODS*_BEAGLE.fits.gz')

        plot(files, catalog=f'{data}/JADES_z6to9LBGcatalog_Endsley2024_f775w_dropouts.fits', dir=f'{figs}/beagle_fits/{name}')
```

```python
plot_beagle_fits()
```
