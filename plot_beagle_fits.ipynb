```python
import sys

import os
import glob

sys.path.append(os.path.abspath('..'))

from mrileyowens.beagle import plot
```

```python
def plot_beagle_fits():

    '''
    Plot the BEAGLE model SED fits, model photometry, and observed photometry
    '''

    # Set common directories
    home = os.getcwd()
    data = f'{home}/data'
    figs = f'{home}/figs'
    results = f'{home}/results'

    # Set the prefixes for the different sets of BEAGLE fits
    names = ['e24_f775w_dropouts_csfh', 'e24_f775w_dropouts_csfh_no_lya', 'e24_f775w_dropouts_2csfh_no_lya']

    # For each set of BEAGLE fits
    for i, name in enumerate(names):

        # Get the BEAGLE fit files
        files = glob.glob(f'{results}/beagle_fits/{name}/*_GOODS*_BEAGLE.fits.gz')

        # Plot the BEAGLE fits
        plot(files, catalog=f'{data}/JADES_z6to9LBGcatalog_Endsley2024_f775w_dropouts.fits', dir=f'{figs}/beagle_fits/{name}', overwrite=True)
```

```python
plot_beagle_fits()
```
